```python
def transform_video_file(file_path, function, params=None, n_frame=-1):
    """
    Given the path of a video file (file_path) the function reads every frame of the input video and applies a given
    transformation (function) using the parameters (params)

    :input_image:        Input video file path
    :function:           Function be applied to each frame of the image. signature `function(np.array, Any) -> Any`
    :params:             Any parameter needed for the function above.
    :n_frame:            Maximum number of frame to read. Default `-1`, read all the content
    :return:             output_handler this can be anything you may need to save your results.
    """
    output_handler = []
    # Open video
    cap = cv.VideoCapture(file_path)
    if cap.isOpened():

        frame_count = 0

        while True:
            if n_frame != -1 and frame_count >= n_frame:
                break

            ret, frame = cap.read()
            if not ret:
                break

            result = function(frame, params)
            output_handler.append(result)

            frame_count += 1

    else:
        raise ValueError('Can not open file: {}'.format(file_path))
    # Close reader
    cap.release()
    # Return custom structure
    return output_handler
```

## 2 Tracking - Particles Filter

A **particles filter** is an object tracking algorithm which try to ***estimate*** the real position of an object approximating its position in the search space (i.e. image space) using a set of particles. A particle has two properties:
 
    - A state: An estimation of the quantity being tracked, in our case it will be a tuple(x, y) that represent the possible location of the object in the image space.
    - A weight: A probability that the particle is at the correct location of the object being tracked
    
The goal of the filter is to estimate the probability density function of the quantity being tracked over the whole search space. The set or particles is an approximation of this probability density function. When there are a lot of particles in a given region, the probability of the object being there is high. 
To estimate this distribution the algorithm uses Monte Carlo techniques and **approximate** this quantity. There is a total of three steps required to implement a particles filter:
 
1. Perturbate: Given a set of particles, the state is slightly perturbed in order to explore the neighbourhood of each particle. This perturbation is generally drawn from a gaussian distribution.
2. Reweight: The particle's weight are updated in this step. Given a set of measurements (*i.e. perturbed particles*) a similarity metric is computed between the measurements and the object model (*i.e. template*). Particle with high similarity will have a large weight (*i.e. the measurement looks very much like the object*) whether a particle with low similarity will be assigned a small weight (*i.e. the measurement does not look like the object*). The reweighting step adapts the estimation of the probability density function of the quantity being tracked.
3. Resample: The set of particles is updated by sampling in the previous collection of particles using the newly estimated probability density function. Since particles similar to the object have a higher chance to be chosen, the new set of particles should have states located on the actual object.
 
In the following section you will implement these three steps from scratch with the help of the class `ParticleFilterInterface` from file `utils.py`. This class has the following members that will be used in the different methods.
 
- `self.model`: The template of the object being tracked, np.ndarray
- `self.search_space`: The dimensions of the search space, tuple
- `self.n_particles`: The total number of particles used to estimate the distribution, int
- `self.state_dims`: The number of component of each state, int (*i.e. 2 in our case, x/y*), tuple
- `self.sigma_perturbation`: Standard deviation of the perturbation applied to the particles, float
- `self.sigma_similarity`: Standard deviation of the residual error used when computing similarity, float
- `self.alpha`: Blending coefficient used during model adaptation phase, float
- `self.particles`: Collection of particles states stored as 2D array, first column is x, second is y, np.ndarray
- `self.weights`: Distribution estimation, 1D array
- `self.state`: Current best estimation of the object position, np.ndarray
- `self.indexes`: List of index for each particles, nd.array

### 2.4 Step Four : Model adaptation (15 pts)

Having a fixed model of the object is not optimal when there is movement (*i.e. out of plane rotation*). Therefore in the next section, a simple model adaptation will be implemented into our particles filter.

The model will not be updated every frame to ensure a smooth transition in term of appearance. The update rate is controled with the `update_rate` parameter in the constructor. The model adaptation consists of a linear interpolation between the patch located at the current best position estimation (*i.e. self.state*) and the previous model.

$$

$$

where $\alpha$ control the amount of blending between the two patches, $P_{best}$ is the patch around the current estimation of the object position and $M_{old}$ is the actual model of the object.

Your task is to complete the implementation of the `detection_fn` which, given an instance of a classifier and an image, return the bounding box where the face is located. More information about the detector can be found in the [doc](https://docs.opencv.org/3.4.4/d1/de5/classcv_1_1CascadeClassifier.html#ab3e572643114c43b21074df48c565a27).

Be sure to implement a solution to deal with the situation where multiple boxes are returned by the classifier. The function must return **ONLY** one bounding box.

When multiple detections are returned, we keep the largest bounding box, assuming it corresponds to the main face in the image.

```python
def detection_fn(detector, image):
    """
    Run face detection on a given `image` with an instance of CascadeClassifier
    
    :param detector: CascadeClassifier instance
    :param image: Frame on which to run detection
    :return: bounding box if any or None
    """
    bbox= None
    
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    faces = detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    if len(faces) > 0:
        # keep only one box: the largest one
        areas = [w * h for (x, y, w, h) in faces]
        idx = np.argmax(areas)
        bbox = tuple(faces[idx])

    return bbox
```

##### 3.4.1.1 Intersection over Union (1pt)

Your task is to complete the implementation of the `iou_metric` function. The function will return the *Intersection over Union* given two bounding boxes, namely `box_a` and `box_b`.

You can find the metric as the Jaccard index https://en.wikipedia.org/wiki/Jaccard_index. Feel free to modify the implementation below to adapt it to your function output.

```python
def iou_metric(box_a, box_b):
    """
    Compute the Intersection over Union (IoU) between two bounding boxes (x, y, w, h)
    
    :param box_a: First bounding box
    :param box_b: Second bounding box
    :return: Intersction over Union
    """
    
    iou = 0.0
    
    x_left_a, y_top_a, w_a, h_a = box_a
    x_left_b, y_top_b, w_b, h_b = box_b
    
    x_right_a = x_left_a + w_a
    y_bottom_a = y_top_a + h_a
    x_right_b = x_left_b + w_b
    y_bottom_b = y_top_b + h_b

    x_left = max(x_left_a, x_left_b)
    y_top = max(y_top_a, y_top_b)
    x_right =  min(x_right_a, x_right_b)
    y_bottom = min(y_bottom_a, y_bottom_b)

    inter_w = max(0, x_right - x_left)

    inter_h = max(0, y_bottom- y_top)

    inter = inter_w*inter_h

    area_a = w_a * h_a
    area_b = w_b * h_b

    union = area_a + area_b - inter

    iou = inter / union

 
    return iou
```

##### 3.4.1.2 Center's distance (1pts)

Complete the `center_metric` function that compute the euclidean distance between two bounding box centers, define as:

$$
d_i = \left|\left| c_i^A - c_i^B \right|\right|
$$

where `A` and `B` are the two bounding boxes (*i.e. detection + ground truth*)


```python
def center_metric(box_a, box_b):
    """
    Compute the Euclidean distance between two given bounding boxes
    
    :param box_a: First bounding box
    :param box_b: Second bounding box
    :return: Distance
    """
    
    d = 0.0
    x_left_a, y_top_a, w_a, h_a = box_a
    x_left_b, y_top_b, w_b, h_b = box_b
    
    x_a =  x_left_a + w_a /2 
    y_a = y_top_a + h_a/2

    x_b =  x_left_b + w_b /2 
    y_b = y_top_b + h_b/2

    d = np.sqrt(np.pow(x_b-x_a,2) + np.pow(y_b-y_a,2))
    
    return d
    
```

##### 3.4.1.3 Execution time (1pt)

The function `time_metric` will measure the execution time of a given function defined in `processing_fn`. Its signature must be `processing_fn() -> Any`.
If the `processing_fn` return some values, they will be passed through the `time_metric` function. Therefore the return value is a tuple containing the execution time in first position and anything returned by `processing_fn` in the second position.

```python
def time_metric(processing_fn):
    """
    Measure the execution time of a given function (i.e. lambda function)
    
    :param processing_fn:  Lambda function to be timed
    :return: Execution time in milli-seconds
    """
    ex_time = 0
    retval = None
    
    start = _time.time()
    retval = processing_fn()
    end = _time.time()

    ex_time = (end - start)*1000
    
    return ex_time, retval
    
```

## Discussion of each tracker

### MIL
`MIL` stands for **Multiple Instance Learning**. The main idea is that, instead of assuming that only one exact image patch corresponds to the object, the tracker considers a small set of nearby patches as possible positive examples. It then learns an appearance model that distinguishes the object from the background. This makes MIL relatively robust when the object location is slightly uncertain, since it does not rely too heavily on a single exact crop. In practice, this usually gives good stability, but it also makes the method slower because the classifier must be updated online and evaluated repeatedly.

**Observed behavior in this experiment:**  
`MIL` performs very well in terms of IoU and distance, but it is relatively slow. It is a strong and stable tracker, though not as efficient as `KCF`, `MedianFlow`, or `MOSSE`.

**Pros**
- robust to moderate ambiguity in object location,
- high accuracy,
- stable over time.

**Cons**
- noticeably slower,
- not the best option for real-time applications.

### KCF
`KCF` stands for **Kernelized Correlation Filter**. It belongs to the family of correlation-filter trackers, which learn a template of the target and then locate it in the next frame by finding where the response is strongest. KCF is particularly efficient because it exploits the properties of cyclic shifts and computes correlations in the Fourier domain. This allows it to evaluate many translated versions of the target very quickly. The “kernelized” part means that it can model more complex similarity relationships than a simple linear filter. In practice, KCF is often one of the best compromises between speed and accuracy, which is also consistent with the experimental results.

**Observed behavior in this experiment:**  
`KCF` is the best overall tracker. It gives the highest IoU, the lowest center distance, and still remains computationally light.

**Pros**
- excellent balance of speed and accuracy,
- low variance,
- stable tracking performance.

**Cons**
- may be less robust under large appearance changes or severe occlusions compared to more specialized long-term methods.

### TLD
`TLD` stands for **Tracking-Learning-Detection**. It combines three components: a short-term tracker, a detector, and an online learning module. The tracker follows the object from one frame to the next, the detector tries to re-detect it if it is lost, and the learning component updates the appearance model over time. The main motivation is long-term tracking: even if the object disappears or the tracker drifts, the detector may bring it back. However, this more complex pipeline can also make the method unstable if the online learning step starts reinforcing wrong examples. That is often why TLD can drift badly or produce erratic results, especially in challenging sequences.

**Observed behavior in this experiment:**  
Despite its more elaborate design, `TLD` performs poorly here. The plots show repeated large drops in IoU and strong spikes in center distance, indicating tracking drift or failure. It is also the slowest tracker in this benchmark.

**Pros**
- designed for long-term tracking with re-detection capability.

**Cons**
- unstable in this sequence,
- high runtime,
- most severe tracking failures among all tested trackers.

  
### MedianFlow
`MedianFlow` is based on tracking many points inside the object region between consecutive frames, usually using optical-flow-like correspondences. These points are tracked forward and backward, and the method checks how consistent those trajectories are. The final motion estimate is then derived from the **median** displacement of the tracked points. Using the median makes the method resistant to outliers, since a few badly tracked points do not dominate the estimate. MedianFlow tends to work well when the motion is smooth and the object appearance does not change too abruptly. On the other hand, if the motion is large or many points become unreliable, the tracker can fail.

**Observed behavior in this experiment:**  
`MedianFlow` performs very well. It has strong IoU, low center distance, and very low runtime. It is one of the best methods overall, slightly behind `KCF` in accuracy but faster.

**Pros**
- good accuracy,
- low processing time,
- robust when motion is smooth and frame-to-frame displacement is moderate.

**Cons**
- may fail when motion is abrupt or when the tracked region changes too much in appearance.


### MOSSE
`MOSSE` stands for **Minimum Output Sum of Squared Error**. It is another correlation-filter tracker, but it is designed to be extremely fast. The method learns a filter that produces a strong response at the target location and weak responses elsewhere. Like KCF, it takes advantage of efficient computations in the frequency domain, but it uses a simpler formulation. Because of this, MOSSE is usually much faster than most other trackers and is well suited for real-time applications. The tradeoff is that it can be less precise or less robust than more sophisticated methods when the object changes scale, rotates strongly, or changes appearance significantly.

**Observed behavior in this experiment:**  
`MOSSE` is by far the fastest method and still keeps a reasonably high IoU. Its center distance is somewhat worse than `KCF` and `MedianFlow`, but still far better than `TLD` and the detection baseline.

**Pros**
- extremely fast,
- good practical accuracy,
- suitable for real-time scenarios.

**Cons**
- slightly less precise than the best trackers,
- may be more sensitive to scale.

### Detection baseline (`Det`)
The baseline uses a **Viola-Jones / CascadeClassifier** style face detector. It scans each frame independently without using any temporal information.

**Observed behavior in this experiment:**  
The detector is clearly less stable than tracking. It can work reasonably well on some frames, but it also produces complete failures and very large localization errors. Its runtime is also higher than several trackers.

**Pros**
- does not depend on the previous frame,
- can re-localize the face even after total loss.

**Cons**
- no temporal consistency,
- unstable across frames,
- weaker average accuracy than the good trackers.


## Pros and cons of the metrics

### IoU
**Pros**
- directly measures overlap between predicted and ground-truth boxes,
- easy to interpret,
- widely used in detection and tracking.

**Cons**
- sensitive to box size and small geometric mismatches,
- two boxes can have similar centers but low IoU if scales differ,
- does not directly describe positional error in pixels.

### Center distance
**Pros**
- intuitive geometric interpretation,
- directly measures localization error,
- useful when position matters more than exact overlap.

**Cons**
- ignores box size,
- two boxes can have a small center distance but poor overlap,
- not normalized, so the meaning depends on image scale.

### Processing time
**Pros**
- essential for practical applications,
- directly measures efficiency and real-time suitability.

**Cons**
- says nothing about accuracy,
- hardware and implementation dependent,

  
These three metrics are complementary: **IoU** captures overlap quality, **center distance** captures localization precision, and **processing time** captures efficiency.



## Final conclusion

The experiment shows that tracking clearly outperforms frame-by-frame detection once the target has been initialized. Among the tested trackers, `KCF` is the strongest overall method because it combines the highest accuracy, lowest localization error, and low runtime. `MedianFlow` is also very competitive and fast, while `MOSSE` is the best choice when speed is the primary concern. `MIL` is accurate but comparatively slow. `TLD` performs poorly on this sequence, with frequent failures and the highest computational cost.

Therefore, for this video and evaluation setup, `KCF` is the best compromise between robustness and efficiency, while `TLD` is the least suitable method.
